```python
import pandas as pd
import os
```

```python
# Ruta del archivo
current_path = os.getcwd()
data_src_path = os.path.join(current_path, "data_src")
file_name = "RQs.csv"
file_path = os.path.join(data_src_path, file_name)
```

```python
file_path
```

```python
# Verificar que el archivo existe
if os.path.exists(file_path):
    # Cargar CSV
    df = pd.read_csv(file_path)
    print("Archivo cargado correctamente. Primeras filas:")
    # print(df.head())
else:
    print(f"Archivo no encontrado en {file_path}")
```

```python
df.head()
```

```python
# Columnas de interés
cols_rq = ["Categories_RQ1", "Categories_RQ2", "Categories_RQ3", "Categories_RQ4"]
```

```python
# Calcular valores distintos y proporciones
for col in cols_rq:
    print(f"\nColumna: {col}")
    value_counts = df[col].value_counts(dropna=False)
    total = len(df)
    for val, count in value_counts.items():
        print(f"  '{val}': {count} de {total} filas ({count/total:.2%})")
```
